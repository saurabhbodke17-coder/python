## Train Test Split
- Train Test Split is used to split the data into training set and testing set
- Training data will be used for the Training model and testing data for model testing
- From your whole dataset you can keep upto 25% Data for testing

```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
data = {
"Study Hours":[1, 1.5, 2, 2.5, 3, 3.5, 4, 4.5, 5],
"Marks":[10,20,30,40,50,60,70,80,90]
}
df = pd.DataFrame(data)
```

```python
x = df[["Study Hours"]]
y = df["Marks"]
```

```python
x_train ,x_test , y_train ,y_test   = train_test_split(x , y , random_state=34 , test_size = 0.33)
```
