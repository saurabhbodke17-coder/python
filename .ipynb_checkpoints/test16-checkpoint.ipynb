```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 103, 105],
    "Name": [
        "Rahul", "Priya", "Amit", "Sneha", "Vikram", "Neha",
        "Arjun", "Pooja", "Rohan", "Anjali", "Amit", "Vikram"
    ],
    "Department": [
        "IT", "HR", "IT", "Finance", "IT", "HR",
        "Finance", "IT", "HR", "Finance", "IT", "IT"
    ],
    "Age": [25, 28, 24, 30, 27, None, 26, 29, 31, 25, 24, 27],
    "Salary": [
        50000, 45000, 55000, 60000, None, 48000,
        65000, 52000, 47000, 58000, 55000, None
    ],
    "City": [
        "Mumbai", "Pune", "Mumbai", "Delhi", "Bangalore", "Pune",
        "Delhi", "Mumbai", "Bangalore", "Delhi", "Mumbai", "Bangalore"
    ],
    "Experience": [2, 3, 1, 5, 4, 3, 6, 2, 7, 2, 1, 4]
}

df = pd.DataFrame(data)

df
```
