```python
# @ above one is about applying only 1 condition
# what if you want to apply multiple conditiiosn?
# &(and) , |(or)
# every condition will be in seperate bracate 
```

```python
# Example: i want to have the people whose experience is greater than 2 years and
# salary greater than 40000

df1 = df[(df["Experience"]>=2) & (df["Salary"] >= 40000)]
df1
# This is how you can write down more than 1 condition=
```

```python
# either experience greater than 1 year or salary greater than 50k 
df2 = df[(df["Experience"]>1) | (df["Salary"]>=50000)] 
df2 
```
